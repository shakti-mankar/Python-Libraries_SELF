### np.sort

Return a sorted copy of an array.

https://numpy.org/doc/stable/reference/generated/numpy.sort.html

```python
# code
import numpy as np
a = np.random.randint(1,100,15)
print(a , "\n")

print(np.sort(a))
```

```python
b = np.random.randint(1,100,24).reshape(6,4)

print(b , "\n")
# print(np.sort(b , "\n"))

print(np.sort(b, axis=1 ))
```

### np.where
