```python
#  ARRAY OPERATIONS 

import numpy as np



arr_1 = np.array([1,2,3,4,5])
arr_2 = np.array([6,7,8,9,1])

add = arr_1+arr_2
minus = arr_1-arr_2
multiply = arr_1*arr_2
mod = arr_1%arr_2
divide = arr_1/arr_2

print(add,minus,multiply,mod, divide,  sep=" " )

```

```python
# DEEP AND SHALLOW COPY 


# Shallow copy 
a = np.arange(1,21)

print(a)

slice = a[:5]
slice = slice *10
print(slice)

print(a)
```

```python
# DEEP COPY

b = a

b[0] = 100

print(a)
```

```python
#  MATRIX OPERATIONS

a = np.array([[1,2], [3,4]])
b = np.array([[5,6],[7,8]])

print(a)

print(b)



print(np.dot(a,b))

```

```python
# TRANSPOSE 

print(a.T)
```

```python
# ADVANCE ARRAY MANIPULATION

# STACKING ARRAY

a = np.array([1,2,3,4])
b = np.array([5,6,7,8])

print(np.vstack((a,b)))

print(np.hstack((a,b)))

print(np.column_stack((a,b)))


```

```python
#  SPLITING ARRAY 

c = np.arange(16).reshape(4,4)


print(np.hsplit(c,4))

print(np.vsplit(c,4))

```
